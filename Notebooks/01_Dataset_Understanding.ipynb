```python
numerical_features = df.select_dtypes(include=np.number).drop(columns=["Machine_ID"])
```

```python
numerical_features.describe()
```

```python
numerical_features.skew()
```

```python
numerical_features.hist(
    figsize=(16, 12),
    bins=30,
    edgecolor="black"
)

plt.suptitle(
    "Distribution of Numerical Features",
    fontsize=16
)

plt.tight_layout()
plt.show()
```

```python
correlation_matrix = numerical_features.corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Features")

plt.show()
```
