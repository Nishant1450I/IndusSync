```python
df["Latency_Category"] = pd.cut(
    df["Network_Latency_ms"],
    bins=[0, 30, 60, float("inf")],
    labels=["Low", "Medium", "High"]
)

print(df["Latency_Category"].value_counts())
```

```python
latency_speed = df.groupby("Latency_Category", observed=True)["Production_Speed_units_per_hr"].mean()

print(latency_speed)
```

```python
latency_speed.plot(kind="bar", figsize=(8, 5))

plt.xlabel("Latency Category")
plt.ylabel("Average Production Speed (units/hr)")
plt.title("Production Speed Under Different Latency Conditions")

plt.show()
```

```python
latency_defects = df.groupby("Latency_Category", observed=True)["Quality_Control_Defect_Rate_%"].mean()

print(latency_defects)
```

```python
latency_defects.plot(kind="bar", figsize=(8, 5))

plt.xlabel("Latency Category")
plt.ylabel("Average Defect Rate (%)")
plt.title("Defect Rate Under Different Latency Conditions")

plt.show()
```

```python
df["Packet_Loss_Category"] = pd.cut(
    df["Packet_Loss_%"],
    bins=[-0.000001, 1, 3, float("inf")],
    labels=["Low", "Medium", "High"]
)

print(df["Packet_Loss_Category"].value_counts())
```

```python
packet_speed = df.groupby("Packet_Loss_Category", observed=True)["Production_Speed_units_per_hr"].mean()

print(packet_speed)
```

```python
packet_speed.plot(kind="bar", figsize=(8, 5))

plt.xlabel("Packet Loss Category")
plt.ylabel("Average Production Speed (units/hr)")
plt.title("Production Speed Under Different Packet Loss Conditions")

plt.show()
```

```python
packet_defects = df.groupby("Packet_Loss_Category", observed=True)["Quality_Control_Defect_Rate_%"].mean()

print(packet_defects)
```

```python
packet_defects.plot(kind="bar", figsize=(8, 5))

plt.xlabel("Packet Loss Category")
plt.ylabel("Average Defect Rate (%)")
plt.title("Defect Rate Under Different Packet Loss Conditions")

plt.show()
```

```python
summary = df.groupby("Latency_Category", observed=True).agg({
    "Network_Latency_ms": "mean",
    "Packet_Loss_%": "mean",
    "Error_Rate_%": "mean",
    "Production_Speed_units_per_hr": "mean",
    "Quality_Control_Defect_Rate_%": "mean",
    "Power_Consumption_kW": "mean"
})

print(summary)
```

```python
packet_summary = df.groupby("Packet_Loss_Category", observed=True).agg({
    "Network_Latency_ms": "mean",
    "Packet_Loss_%": "mean",
    "Error_Rate_%": "mean",
    "Production_Speed_units_per_hr": "mean",
    "Quality_Control_Defect_Rate_%": "mean",
    "Power_Consumption_kW": "mean"
})

print(packet_summary)
```

```python
comparison = pd.DataFrame({
    "Latency": latency_speed,
    "Packet Loss": packet_speed
})

print(comparison)
```

```python
summary.to_csv("network_latency_summary.csv")
packet_summary.to_csv("network_packet_loss_summary.csv")

print("Network analysis summary files saved successfully!")
```
